```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./data/titanic_clean.csv')
```

```python
df.sample(4)
```

```python
df['Survived'].value_counts()
```

```python
plt.figure(figsize=(12, 8))

sns.barplot(x = df['Survived'].value_counts().index, y = df['Survived'].value_counts())
plt.title('Value Counts of Survived')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize = (12, 8))

# colores para cada una de las clases
color = ['#240750', '#344C64']

sns.countplot(df, x = 'Embarked', hue = 'Survived', palette=color) 
plt.legend(title = 'Supervivencia', labels = ['No sobrevivió', 'Sí sobrevivió'])
plt.show()
```

```python
plt.figure(figsize = (12, 8))

sns.histplot(df, x = 'Age', kde = True)
plt.title('Distribución de Edades')
plt.axvline(df['Age'].mean(), color = 'red')
plt.axvline(df['Age'].median(), color = 'green')
plt.axvline(df['Age'].mode()[0], color = 'black')
plt.show()

# Print the results
print(f"Media: {df['Age'].mean()}")
print(f"Mediana: {df['Age'].median()}")
print(f"Moda: {df['Age'].mode()[0]}")
```

```python
plt.figure(figsize = (12, 8))
sns.heatmap(df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].corr(), 
            annot=True, 
            cbar=True,  # Activar la barra de colores
            cmap='coolwarm',  # Cambiar la paleta de colores
            cbar_kws={'label': 'Nivel de correlación'})  # Etiqueta para la barra de colores
plt.title('Correlación de variables numéricas')
plt.show()
```

```python

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Age', kde=True, color='skyblue', bins=30, edgecolor='black')

plt.title('Distribución de Edades de los Pasajeros del Titanic', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Edad (Años)', fontsize=12)
plt.ylabel('Frecuencia (Número de Pasajeros)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, x='Pclass', hue='Survived', palette='Set2', edgecolor='black')

plt.title('Cantidad de Sobrevivientes y No Sobrevivientes por Clase de Pasajero', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Clase de Pasajero (Pclass)', fontsize=12)
plt.ylabel('Cantidad de Pasajeros', fontsize=12)
plt.legend(title='Estado', labels=['No Sobrevivió', 'Sobrevivió'])

ax.set_xticklabels(['1ª Clase', '2ª Clase', '3ª Clase'])

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, x='Sex', hue='Survived', palette='pastel', edgecolor='black')

plt.title('Cantidad de Sobrevivientes y No Sobrevivientes por Género', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Cantidad de Pasajeros', fontsize=12)

ax.set_xticklabels(['Masculino (male)', 'Femenino (female)'])
plt.legend(title='Estado', labels=['No Sobrevivió', 'Sobrevivió'])

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, x='Embarked', hue='Survived', palette='Set1', edgecolor='black')

plt.title('Cantidad de Sobrevivientes y No Sobrevivientes por Puerto de Embarque', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Puerto de Embarque', fontsize=12)
plt.ylabel('Cantidad de Pasajeros', fontsize=12)

ax.set_xticklabels(['Southampton (S)', 'Cherbourg (C)', 'Queenstown (Q)'])
plt.legend(title='Estado', labels=['No Sobrevivió', 'Sobrevivió'])

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

bins = [0, 2, 12, 18, 35, 60, 100]
labels = ['Infante', 'Niño', 'Adolescente', 'Joven Adulto', 'Adulto', 'Mayor']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, include_lowest=True)

survival_rates = df.groupby('AgeGroup', observed=False)['Survived'].mean().reset_index()
plt.figure(figsize=(10, 6))
ax = sns.barplot(data=survival_rates, x='AgeGroup', y='Survived', palette='vlag', edgecolor='black')

plt.title('Tasa de Supervivencia por Grupo de Edad', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Grupo de Edad', fontsize=12)
plt.ylabel('Tasa de Supervivencia (Proporción)', fontsize=12)
plt.ylim(0, 1)

for p in ax.patches:
    ax.annotate(f'{p.get_height():.2f}', 
                (p.get_x() + p.get_width() / 2., p.get_height() + 0.02), 
                ha='center', va='center', fontsize=11, fontweight='bold', color='black')

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(9, 5))
ax = sns.countplot(data=df, x='Pclass', palette='Set2', edgecolor='black')

plt.title('Distribución Total de Pasajeros por Clase', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Clase de Pasajero (Pclass)', fontsize=12)
plt.ylabel('Cantidad de Pasajeros', fontsize=12)

ax.set_xticklabels(['1ª Clase', '2ª Clase', '3ª Clase'])

for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height() + 5), 
                ha='center', va='center', fontsize=11, fontweight='bold', color='black')

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

media_edad = df['Age'].mean()
mediana_edad = df['Age'].median()
moda_edad = df['Age'].mode().iloc[0] 

plt.figure(figsize=(11, 6))
sns.histplot(data=df, x='Age', kde=True, color='skyblue', bins=30, edgecolor='black', alpha=0.7)

plt.axvline(media_edad, color='red', linestyle='--', linewidth=2, 
            label=f'Media: {media_edad:.2f} años')
plt.axvline(mediana_edad, color='green', linestyle='-.', linewidth=2, 
            label=f'Mediana: {mediana_edad:.2f} años')
plt.axvline(moda_edad, color='purple', linestyle=':', linewidth=2.5, 
            label=f'Moda: {moda_edad:.2f} años')

plt.title('Distribución de Edades con Media, Mediana y Moda', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Edad (Años)', fontsize=12)
plt.ylabel('Frecuencia (Número de Pasajeros)', fontsize=12)

plt.legend(fontsize=11, loc='upper right', facecolor='white', edgecolor='gray')

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 5))
ax = sns.boxplot(data=df, x='Age', color='lightcoral', flierprops={"marker": "o", "markerfacecolor": "red", "markersize": 8})

plt.title('Diagrama de Caja de Edades para Identificación de Outliers', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Edad (Años)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

tarifa_maxima = df['Fare'].max()
tarifa_minima = df['Fare'].min()

print(f"La tarifa máxima pagada fue: ${tarifa_maxima:.2f}")
print(f"La tarifa mínima pagada fue: ${tarifa_minima:.2f}")
```

```python
sns.set_theme(style="whitegrid")

media_tarifa = df['Fare'].mean()
mediana_tarifa = df['Fare'].median()

moda_tarifa = df['Fare'].mode().iloc[0]

plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='Fare', kde=True, color='teal', bins=40, edgecolor='black', alpha=0.6)

plt.axvline(media_tarifa, color='red', linestyle='--', linewidth=2, 
            label=f'Media: ${media_tarifa:.2f}')
plt.axvline(mediana_tarifa, color='green', linestyle='-.', linewidth=2, 
            label=f'Mediana: ${mediana_tarifa:.2f}')
plt.axvline(moda_tarifa, color='purple', linestyle=':', linewidth=2.5, 
            label=f'Moda: ${moda_tarifa:.2f}')

plt.title('Distribución de Tarifas con Media, Mediana y Moda', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Tarifa (Fare)', fontsize=12)
plt.ylabel('Frecuencia (Número de Pasajeros)', fontsize=12)

plt.legend(fontsize=11, loc='upper right', facecolor='white', edgecolor='gray')
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(12, 5))
ax = sns.boxplot(data=df, x='Fare', color='mediumaquamarine', 
                 flierprops={"marker": "o", "markerfacecolor": "crimson", "markersize": 6, "markeredgecolor": "black"})

plt.title('Diagrama de Caja de Tarifas para Identificación de Outliers', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Tarifa (Fare)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
porcentaje_sibsp = (
    df.groupby('SibSp')['Survived']
    .mean()
    .reset_index()
)
porcentaje_sibsp['Survived'] = porcentaje_sibsp['Survived'] * 100
porcentaje_sibsp = porcentaje_sibsp.rename(columns={'Survived': 'Porcentaje de Supervivencia (%)'})

print("Porcentaje de supervivencia según hermanos/esposos a bordo:")
porcentaje_sibsp.style.format({'Porcentaje de Supervivencia (%)': '{:.2f}%'}).hide()
```

```python
sns.set_theme(style="white")

columnas_numericas = df.select_dtypes(include=['number'])
matriz_correlacion = columnas_numericas.corr()


plt.figure(figsize=(10, 8))

sns.heatmap(matriz_correlacion, 
            annot=True,       
            fmt=".2f",        
            cmap='coolwarm',  
            vmax=1, vmin=-1,   
            center=0,          
            linewidths=0.5,     
            cbar_kws={"shrink": .8}) 

plt.title('Matriz de Correlación de Variables Numéricas', fontsize=15, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```
